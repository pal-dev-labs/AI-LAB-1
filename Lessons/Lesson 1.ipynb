## Operatori aritmetici

| Operatore    | Nome           | Descrizione                                            |
|--------------|----------------|--------------------------------------------------------|
| ``a + b``    | Addizione      | Somma di ``a`` e ``b``                                 |
| ``a - b``    | Sottrazione    | Differenza tra ``a`` e ``b``                           |

Gli operatori seguono le proprie **regole di precedenza** (simili a quelle della matematica). Per modificare tali precedenze, si utilizzano le parentesi. Una differenza importante è che in Python, a questo scopo, si impiegano **esclusivamente le parentesi tonde**: è possibile aprirne quante se ne desidera, una dentro l'altra; l'importante è ricordarsi di chiuderle! Dimenticarsi di chiudere una parentesi aperta è un errore piuttosto comune.

| **Matematica** | **Python** | |

| $(\left( \frac{1}{4} + 2 \left[ \frac{1}{5} - \left( 1 + \frac{1}{3} \right) \right] \right)^2 )$   | `(1 / 4 + 2 * (1 / 5 - (1 + 1 / 3))) ** 2` | Tre parentesi tonde aperte e tre chiuse. |
| Una piattaforma circolare ruota di un angolo di 165° per 12 volte consecutive. * **OUTPUT / STAMPA:** Visualizzazione del risultato sullo schermo (`100`).
* **FINE:** Termine del programma.

# ESERCIZIO: MOSTRA ESITO DI UNA VERIFICA 

Scrivi un programma che chieda all'insegnante di inserire il **voto ottenuto da uno studente** in una prova scritta (un numero decimale da `1` a `10`) e comunichi a schermo:
* `"Esito: SUFFICIENTE"` se il voto è maggiore o uguale a `6`;
* `"Esito: INSUFFICIENTE - Recupero necessario"` in caso contrario (se il voto è strettamente minore di `6`).

---

## Fase 1: Analisi IPO (Input - Processing - Output)

| Fase | Nome Variabile | Descrizione | Tipo di dato |
| :--- | :--- | :--- | :--- |
| **Input** | `voto` | Valutazione numerica della prova scritta (1–10) | `float` |
| **Elaborazione** | *(Condizione)* | Verifica SE `voto >= 6` (Vero / Falso) | `bool` |
| **Output** | *(Messaggio)* | Stringa di testo con l'esito della verifica | `str` |

---

## Fase 2: Progettazione in Pseudocodice

```text
ALGORITMO Verifica_Sufficienza

DICHIARAZIONE VARIABILI:
    Prepara uno spazio di memoria per numero decimale (REALE): voto

INIZIO
    // 1. Fase di Input
    Mostra messaggio: "Inserisci il voto dello studente (1-10):"
    Ricevi il dato e salvalo nella variabile "voto"

    // 2. Fase di Decisione / Elaborazione (Selezione Binaria)
    SE (voto >= 6) ALLORA
        Mostra messaggio: "Esito: SUFFICIENTE"
    ALTRIMENTI
        Mostra messaggio: "Esito: INSUFFICIENTE - Recupero necessario"
    FINE SE
FINE
```

## Fase 3: Diagramma di Flusso (Flowchart)
Nei diagrammi di flusso, la scelta si rappresenta con un **rombo** da cui escono due rami alternativi (`Vero / Sì` e `Falso / No`) che si ricongiungono prima della fine dell'algoritmo:

    C -- Vero (Sì) --> D[/Stampa: Esito SUFFICIENTE/]

---

## Fase 4: Implementazione in Python
*Nella cella di codice sottostante (`Code`), implementa l'algoritmo rispettando l'indentazione (4 spazi) dopo i due punti `:`.*

```python
# 1. Fase di Input
voto = float(input("Inserisci il voto dello studente (1-10): "))

# 2. Fase di Selezione (if / else)
if voto >= 6:
    print("Esito: SUFFICIENTE")
else:
    print("Esito: INSUFFICIENTE - Recupero necessario")
```

---

## Note didattiche: dalla Sequenza alla Selezione
* **Mutua esclusività:** A differenza della programmazione sequenziale (dove vengono eseguite *tutte* le istruzioni dall'alto verso il basso), qui i due rami sono **alternativi**: il computer esegue o il blocco `if` oppure il blocco `else`, mai entrambi.
* **L'indentazione (rientro del testo):** In Python i due punti (`:`) al termine di `if voto >= 6:` e `else:` obbligano a rientrare le istruzioni successive di 4 spazi; questo rientro delimita visivamente e sintatticamente ciò che appartiene a ciascun ramo della scelta.
* **Operatore relazionale `>=` vs `>`:** Prestare attenzione ai valori limite: scrivendo `voto > 6`, il voto `6` pieno verrebbe classificato erroneamente come insufficiente.

## LE STRUTTURE CONDIZIONALI IN PYTHON

## Istruzioni condizionali: ``if``-``elif``-``else``:
Le istruzioni condizionali, spesso chiamate istruzioni *if-then* (se-allora), consentono al programmatore di eseguire determinati blocchi di codice in base a una condizione booleana.
Un esempio basilare di istruzione condizionale in Python è il seguente:

````python
x = 1
if x > 3:    # se il valore boolean è 3
    print("number is bigger than 3")
````

````python
x = 0

if x >= 0:
    print("X is bigger than 0")
else:
    print(x, "x is lower than 0")
````

Possiamo considerare più alternative

````python
x = float(input("insert a number:"))

if x == 0:
     print("the number entered is zero")
elif x < 0:
     print("the number entered is negative")
elif (0 < x) and (x <= 5):
     print("the number is between (or equal to) 0 and 5")
else:
     print("the number is greater than 5")
````

L'oggetto lista può essere creato utilizzando l'operatore **[]**.

Possiamo anche utilizzare variabili all'interno dell'operatore []

````python
a = 10
b="Hello"
squares=[1,4,9,16,25,36]
list2 = [a,b,squares]   
print(list2)
````

````python
# Le liste possono essere indicizzate e suddivise tramite slicing:
print(squares)
print(squares[0])  # l'indicizzazione restituisce l'elemento
print(squares[0:3]) # lo slicing restituisce una nuova lista
print(squares[-1]) # parte dall'ultimo elemento
print(squares[1:-1]) # stampa escludendo il primo e l'ultimo elemento

# Tutte le operazioni di slicing restituiscono una nuova lista contenente gli elementi richiesti.
````

````python
# Le liste supportano anche operazioni come la concatenazione:
print(squares + [36, 49, 64, 81, 100])
````

````python
# A differenza delle stringhe, che sono immutabili, le liste sono un tipo mutabile;
# ciò significa che è possibile modificarne il contenuto:
cubes = [1, 8, 27, 65, 125]  # c'è qualcosa che non va: il cubo di 4 è 64!
cubes[3] = 64  # sostituiamo il valore errato
print( cubes )
````
